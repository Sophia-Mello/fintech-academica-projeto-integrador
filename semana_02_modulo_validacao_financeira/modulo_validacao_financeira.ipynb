# Semana 2 — Módulo de validação financeira

**Objetivo:** receber despesas estudantis e impedir o registro de informações inválidas. Nesta etapa, os registros ficam em memória; a integração com o SQLite será feita nas atividades seguintes.

**Regras:** valor numérico maior que zero e finito, data como `datetime` e categoria de uma lista definida.


```python
from datetime import datetime
from math import isfinite

CATEGORIAS_VALIDAS = ('Alimentação','Transporte','Material Didático','Moradia','Lazer','Outros')

def validar_despesa(despesa):
    if not isinstance(despesa,dict):
        return False,'Informe um dicionário com os dados da despesa.'
    obrigatorios=('valor','data','categoria')
    faltantes=[chave for chave in obrigatorios if chave not in despesa]
    if faltantes:
        return False,'Campo(s) ausente(s): '+', '.join(faltantes)
    valor=despesa['valor']
    if isinstance(valor,bool) or not isinstance(valor,(int,float)) or not isfinite(valor) or valor<=0:
        return False,'O valor deve ser numérico, finito e maior que zero.'
    if not isinstance(despesa['data'],datetime):
        return False,'A data deve ser informada como datetime.'
    if despesa['categoria'] not in CATEGORIAS_VALIDAS:
        return False,'Categoria não permitida.'
    return True,'Despesa validada.'

despesas=[]

def adicionar_despesa(dados):
    valido,mensagem=validar_despesa(dados)
    if valido:
        despesas.append(dados.copy())
    return valido,mensagem
```

## Casos de teste

Além das entradas válidas, testo zero, número negativo, texto no valor, data em formato incorreto, categoria desconhecida e campos ausentes.


```python
casos = [
    ('Alimentação válida', {'valor':35.90,'data':datetime(2026,9,1),'categoria':'Alimentação'}, True),
    ('Transporte válido', {'valor':12.50,'data':datetime(2026,9,2),'categoria':'Transporte'}, True),
    ('Valor negativo', {'valor':-12,'data':datetime(2026,9,2),'categoria':'Lazer'}, False),
    ('Valor zero', {'valor':0,'data':datetime(2026,9,2),'categoria':'Lazer'}, False),
    ('Texto no valor', {'valor':'80','data':datetime(2026,9,2),'categoria':'Lazer'}, False),
    ('Valor infinito', {'valor':float('inf'),'data':datetime(2026,9,2),'categoria':'Lazer'}, False),
    ('Data incorreta', {'valor':80,'data':'2026-09-02','categoria':'Lazer'}, False),
    ('Categoria inválida', {'valor':80,'data':datetime(2026,9,2),'categoria':'Combustível'}, False),
    ('Campo ausente', {'valor':80,'categoria':'Lazer'}, False),
]
for descricao,dados,esperado in casos:
    resultado,mensagem=adicionar_despesa(dados)
    assert resultado==esperado,descricao
    print(f'{descricao:20s} | {resultado!s:5s} | {mensagem}')
assert len(despesas)==2
print('Quantidade de despesas aceitas:',len(despesas))
```

## Conclusão

A validação foi aplicada antes de adicionar os dados à lista. Assim, casos inválidos não entram no total das despesas. O próximo passo é organizar o código em classes para separar as responsabilidades.

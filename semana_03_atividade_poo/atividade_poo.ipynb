# Semana 3 — Programação Orientada a Objetos

Retomei as validações da semana anterior e dividi o sistema em três classes: `Estudante`, `Despesa` e `GestorFinanceiro`. Isso permite associar cada lançamento a um estudante e centralizar as operações de soma.


```python
from datetime import datetime
from math import isfinite

CATEGORIAS = ('Alimentação','Transporte','Material Didático','Moradia','Lazer','Outros')

class Estudante:
    def __init__(self, nome, matricula):
        if not isinstance(nome,str) or not nome.strip():
            raise ValueError('Nome obrigatório.')
        if not isinstance(matricula,str) or not matricula.strip():
            raise ValueError('Matrícula obrigatória.')
        self.nome=nome.strip()
        self.matricula=matricula.strip()

class Despesa:
    def __init__(self, valor, categoria, data):
        self.__valor=None
        self.__categoria=None
        if not isinstance(data,datetime):
            raise ValueError('Data inválida.')
        self.data=data
        self.set_valor(valor)
        self.set_categoria(categoria)

    def set_valor(self,valor):
        if isinstance(valor,bool) or not isinstance(valor,(int,float)) or not isfinite(valor) or valor<=0:
            raise ValueError('Valor inválido.')
        self.__valor=float(valor)

    def get_valor(self):
        return self.__valor

    def set_categoria(self,categoria):
        if categoria not in CATEGORIAS:
            raise ValueError('Categoria inválida.')
        self.__categoria=categoria

    def get_categoria(self):
        return self.__categoria

class GestorFinanceiro:
    def __init__(self,estudante):
        if not isinstance(estudante,Estudante):
            raise TypeError('É necessário fornecer um Estudante.')
        self.estudante=estudante
        self.__despesas=[]

    def adicionar_despesa(self,despesa):
        if not isinstance(despesa,Despesa):
            raise TypeError('Somente objetos Despesa são aceitos.')
        self.__despesas.append(despesa)

    def calcular_total(self):
        return sum(x.get_valor() for x in self.__despesas)

    def relatorio(self):
        return {'estudante':self.estudante.nome, 'matricula':self.estudante.matricula,
                'quantidade':len(self.__despesas),'total':self.calcular_total()}
```

## Exercício de consolidação e testes


```python
aluno=Estudante('João da Silva','2026001')
gestor=GestorFinanceiro(aluno)
for valor,categoria in [(35.90,'Alimentação'),(12.50,'Transporte'),(89.90,'Material Didático')]:
    gestor.adicionar_despesa(Despesa(valor,categoria,datetime(2026,9,3)))
assert round(gestor.calcular_total(),2)==138.30
print(gestor.relatorio())

for valor,categoria in [(0,'Lazer'),(-3,'Lazer'),('30','Lazer'),(50,'Categoria inexistente')]:
    try:
        Despesa(valor,categoria,datetime(2026,9,3))
        raise AssertionError('Este caso deveria ser rejeitado.')
    except ValueError:
        pass
print('Testes de validação concluídos.')
```

## Conclusão

A classe `Despesa` valida os atributos ao criar ou alterar um objeto. A lista de despesas fica encapsulada no gestor, que calcula o total sem permitir a inclusão de outros tipos de objeto. A persistência ainda é em memória nesta semana.

# Projeto de Gestão de Estoque - Semana 1

## Objetivo

O objetivo deste projeto é representar e analisar o estoque de uma empresa.
O código organiza informações dos produtos, como id, nome, categoria, preço e quantidade, permitindo realizar análises sobre o estoque.

## Metodologia

Foi utilizada a biblioteca Pandas para transformar uma lista de produtos em um DataFrame.

O DataFrame facilita a organização, visualização e análise dos dados, permitindo realizar operações como filtros, cálculos, agrupamentos e comparação de valores entre as categorias dos produtos.

```python
import pandas as pd
```

```python
produtos = [
    {"id": 1, "nome": "Notebook", "categoria": "Eletrônicos", "preco": 3500.00, "quantidade": 8},
    {"id": 2, "nome": "Mouse", "categoria": "Acessórios", "preco": 80.00, "quantidade": 25},
    {"id": 3, "nome": "Teclado", "categoria": "Acessórios", "preco": 150.00, "quantidade": 15},
    {"id": 4, "nome": "Monitor", "categoria": "Eletrônicos", "preco": 1200.00, "quantidade": 10},
    {"id": 5, "nome": "Cadeira Gamer", "categoria": "Móveis", "preco": 900.00, "quantidade": 6},
    {"id": 6, "nome": "Headset", "categoria": "Acessórios", "preco": 220.00, "quantidade": 12},
    {"id": 7, "nome": "Webcam", "categoria": "Eletrônicos", "preco": 300.00, "quantidade": 9},
    {"id": 8, "nome": "Impressora", "categoria": "Eletrônicos", "preco": 850.00, "quantidade": 5},
    {"id": 9, "nome": "Mesa de Escritório", "categoria": "Móveis", "preco": 650.00, "quantidade": 7},
    {"id": 10, "nome": "Caixa de Som", "categoria": "Áudio", "preco": 180.00, "quantidade": 14}
]
```

## Criando o DataFrame

Agora vamos transformar a lista de produtos em um DataFrame do Pandas.

```python
df = pd.DataFrame(produtos)

df
```

## Pergunta 1: Qual é o produto mais caro?

```python
produto_mais_caro = df.loc[df["preco"].idxmax()]

produto_mais_caro
```

**Resposta:** O produto mais caro é o Notebook, com preço de R$ 3.500,00.

## Pergunta 2: Qual produto possui a maior quantidade em estoque?

```python
maior_estoque = df.loc[df["quantidade"].idxmax()]

maior_estoque
```

**Resposta:** O Mouse possui a maior quantidade em estoque, com 25 unidades.

## Pergunta 3: Qual é o valor total de cada produto em estoque?

Para descobrir o valor em estoque, multiplicamos o preço pela quantidade.

```python
df["valor_estoque"] = df["preco"] * df["quantidade"]

df
```

## Pergunta 4: Qual é o valor total de todo o estoque?

```python
valor_total = df["valor_estoque"].sum()

print(f"Valor total do estoque: R$ {valor_total:.2f}")
```

## Pergunta 5: Qual categoria possui o maior valor em estoque?

```python
valor_por_categoria = (
    df.groupby("categoria")["valor_estoque"]
    .sum()
    .sort_values(ascending=False)
)

valor_por_categoria
```

**Resposta:** A categoria Eletrônicos possui o maior valor em estoque.

## Pergunta 6: Quais produtos estão com estoque baixo?

Vamos considerar estoque baixo quando o produto possui menos de 10 unidades.

```python
estoque_baixo = df[df["quantidade"] < 10]

estoque_baixo[["nome", "quantidade"]]
```

**Resposta:** Os produtos com menos de 10 unidades são:

Notebook, Cadeira Gamer, Webcam, Impressora e Mesa de Escritório.

## Pergunta 7: Qual é o preço médio dos produtos por categoria?

```python
preco_medio_categoria = (
    df.groupby("categoria")["preco"]
    .mean()
    .sort_values(ascending=False)
)

preco_medio_categoria
```

## Pergunta 8: Quantos produtos existem em cada categoria?

```python
quantidade_por_categoria = df.groupby("categoria")["id"].count()

quantidade_por_categoria
```

## Pergunta 9: Qual é o produto mais barato?

```python
produto_mais_barato = df.loc[df["preco"].idxmin()]

produto_mais_barato
```

**Resposta:** O produto mais barato é o Mouse, com preço de R$ 80,00.

## Pergunta 10: Qual é a quantidade total de produtos em estoque?

```python
quantidade_total = df["quantidade"].sum()

print("Quantidade total de itens em estoque:", quantidade_total)
```

## Conclusão

Após analisar os dados do estoque, foi possível identificar que a categoria Eletrônicos possui o maior valor agregado em estoque, totalizando R$ 46.950,00.
